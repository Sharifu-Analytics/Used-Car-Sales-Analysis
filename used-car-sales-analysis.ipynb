```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# -----------------------------
# KPI values
# -----------------------------
total_cars = len(df)
average_price = df["car_price"].mean()
median_price = df["car_price"].median()
average_mileage = df["car_mileage"].mean()

# -----------------------------
# Data for charts
# -----------------------------
top_brands = df["car_brand"].value_counts().head(10)
country_distribution = df["car_country"].value_counts().head(10)

fuel_price = (
    df.groupby("car_fuel")["car_price"]
      .mean()
      .sort_values(ascending=False)
)

transmission_price = (
    df.groupby("car_transmission")["car_price"]
      .mean()
      .sort_values(ascending=False)
)

# Ruble formatter
def ruble_format(value, position):
    return f"₽{value:,.0f}"

# -----------------------------
# Create 16:9 dashboard
# -----------------------------
fig = plt.figure(figsize=(16, 9))

fig.suptitle(
    "USED CAR SALES DASHBOARD",
    fontsize=20,
    fontweight="bold",
    y=0.97
)

# -----------------------------
# KPI cards
# -----------------------------
card_style = dict(
    edgecolor="black",
    boxstyle="round,pad=0.7",
    linewidth=1.2
)

fig.text(
    0.12, 0.89,
    f"Total Cars\n{total_cars:,}",
    ha="center",
    va="center",
    fontsize=11,
    fontweight="bold",
    bbox=dict(facecolor="skyblue", **card_style)
)

fig.text(
    0.37, 0.89,
    f"Average Price\n₽{average_price:,.0f}",
    ha="center",
    va="center",
    fontsize=11,
    fontweight="bold",
    bbox=dict(facecolor="lightgreen", **card_style)
)

fig.text(
    0.62, 0.89,
    f"Median Price\n₽{median_price:,.0f}",
    ha="center",
    va="center",
    fontsize=11,
    fontweight="bold",
    bbox=dict(facecolor="khaki", **card_style)
)

fig.text(
    0.87, 0.89,
    f"Average Mileage\n{average_mileage:,.0f} km",
    ha="center",
    va="center",
    fontsize=11,
    fontweight="bold",
    bbox=dict(facecolor="lightcoral", **card_style)
)

# -----------------------------
# Compact dashboard layout
# -----------------------------
ax1 = plt.subplot(2, 3, 1)
ax2 = plt.subplot(2, 3, 2)
ax3 = plt.subplot(2, 3, 3)
ax4 = plt.subplot(2, 3, 4)
ax5 = plt.subplot(2, 3, 5)
ax6 = plt.subplot(2, 3, 6)

# 1. Price Distribution
ax1.hist(df["car_price"], bins=40)
ax1.set_xlim(0, 10_000_000)
ax1.set_title("Price Distribution", fontsize=11)
ax1.set_xlabel("Car Price", fontsize=9)
ax1.set_ylabel("Frequency", fontsize=9)
ax1.xaxis.set_major_formatter(FuncFormatter(ruble_format))
ax1.tick_params(axis="both", labelsize=8)

# 2. Price vs Mileage
ax2.scatter(
    df["car_mileage"],
    df["car_price"],
    alpha=0.18,
    s=5
)
ax2.set_title("Car Price vs Mileage", fontsize=11)
ax2.set_xlabel("Mileage (km)", fontsize=9)
ax2.set_ylabel("Car Price", fontsize=9)
ax2.yaxis.set_major_formatter(FuncFormatter(ruble_format))
ax2.tick_params(axis="both", labelsize=8)

# 3. Top 10 Brands
top_brands.plot(kind="bar", ax=ax3)
ax3.set_title("Top 10 Car Brands", fontsize=11)
ax3.set_xlabel("Car Brand", fontsize=9)
ax3.set_ylabel("Number of Cars", fontsize=9)
ax3.tick_params(axis="x", rotation=45, labelsize=8)
ax3.tick_params(axis="y", labelsize=8)

# 4. Top 10 Countries
country_distribution.plot(kind="bar", ax=ax4)
ax4.set_title("Top 10 Countries", fontsize=11)
ax4.set_xlabel("Country", fontsize=9)
ax4.set_ylabel("Number of Cars", fontsize=9)
ax4.tick_params(axis="x", rotation=45, labelsize=8)
ax4.tick_params(axis="y", labelsize=8)

# 5. Fuel Type
fuel_price.plot(kind="bar", ax=ax5)
ax5.set_title("Average Price by Fuel Type", fontsize=11)
ax5.set_xlabel("Fuel Type", fontsize=9)
ax5.set_ylabel("Average Price", fontsize=9)
ax5.yaxis.set_major_formatter(FuncFormatter(ruble_format))
ax5.tick_params(axis="x", rotation=30, labelsize=8)
ax5.tick_params(axis="y", labelsize=8)

# 6. Transmission
transmission_price.plot(kind="bar", ax=ax6)
ax6.set_title("Average Price by Transmission", fontsize=11)
ax6.set_xlabel("Transmission", fontsize=9)
ax6.set_ylabel("Average Price", fontsize=9)
ax6.yaxis.set_major_formatter(FuncFormatter(ruble_format))
ax6.tick_params(axis="x", rotation=30, labelsize=8)
ax6.tick_params(axis="y", labelsize=8)

# Footer
fig.text(
    0.5,
    0.015,
    "Created by Md Shariful Islam | Python • Pandas • Matplotlib",
    ha="center",
    fontsize=9
)

# Leave space for title and KPI cards
plt.tight_layout(rect=[0.03, 0.05, 0.97, 0.82])

# Save high-quality image
plt.savefig(
    "Used_Car_Sales_Dashboard_16x9.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
